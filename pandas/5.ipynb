```python
import pandas as pd
df = pd.read_csv("https://cognitrilha.com.br/datasets/demostrativo_acidentes_eco101.csv.gz", encoding="latin1", sep=";")
df
```

```python
df["sentido"].unique()
```

```python
df = df.dropna(subset="sentido")
```

```python
df["sentido"].unique()
```

```python
df.loc[df["sentido"]=="Crescente","sentido"] = "Sul"
df
```

```python
df.loc[df["sentido"]=="Decrescente","sentido"] = "Norte"
df
```

Crie uma coluna "total_vitimas", como a soma das demais colunas de categorias de vítimas.

```python
df["total_vitimas"] = df["levemente_feridos"] + df["moderadamente_feridos"] + df["gravemente_feridos"] + df["mortos"]
df
```

Liste os valores únicos da coluna tipo_de_ocorrencia.


```python
df["tipo_de_ocorrencia"].unique()
```

Filtre apenas ocorrências com total_vitimas > 0.

```python
df[df["total_vitimas"] > 0]
```

```python
df.columns
```


Quantos acidentes com vítimas envolveram caminhões?

```python
df[df["caminhao"] >0]
```

Qual o acidente com o maior número de vítimas totais?

```python
df.sort_values("total_vitimas").max()
```

```python
df["data"] = pd.to_datetime(df["data"], format='%d/%m/%Y')
df
```

```python
df["mes"] = df["data"].dt.month
df
```

```python
df["ano"] = df["data"].dt.year
df
```

Mostre em um gráfico de barras o total de vítimas de cada ano.

```python
df.groupby("ano")["total_vitimas"].sum().plot.bar()
```

Mostre em um gráfico de barras a média de vítimas agrupadas por mês.


```python
df.groupby("mes")["total_vitimas"].mean().plot.bar()
```

Mostre em um gráfico de pizza o percentual dos acidentes em cada sentido da via, considerando apenas os acidentes que possuem mortos maior que zero

```python
df[df["mortos"]>0].groupby("sentido").size().plot.pie()
```

Exiba dois gráficos do tipo boxplot: um com os dados do total de vítimas e outro com os dados do total de mortes

```python
df.boxplot(column=["total_vitimas","mortos"])
```
